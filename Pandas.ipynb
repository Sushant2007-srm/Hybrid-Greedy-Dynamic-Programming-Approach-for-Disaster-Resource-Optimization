```python
#Selection by Label:

#We can select any row by passing the row label to a loc function.
import pandas as pd
#Imports the Pandas library and gives it the alias 'pd'
#AP24110010909

info = {'one' : pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),
'two' : pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}
#Creates a dictionary containing two Pandas Series with their respective indexes
#AP24110010909

df = pd.DataFrame(info)
#Converts the dictionary of Series into a Pandas DataFrame
#AP24110010909

print("Data frame:")
#Displays a message before printing the DataFrame
#AP24110010909

print(df)
#Displays the DataFrame on the screen
#AP24110010909

print("b index elements are:")
#Displays a message indicating that the row with index 'b' will be selected
#AP24110010909

print (df.loc['b'])
#Selects and displays the row with the label 'b' using the loc function
#AP24110010909

print("e index elements are:")
#Displays a message indicating that the row with index 'e' will be selected
#AP24110010909

print(df.loc['e'])
#Selects and displays the row with the label 'e' using the loc function
#AP24110010909
```

```python
#Addition of rows:

import pandas as pd
#Imports the Pandas library and gives it the alias 'pd'
#AP24110010909

#Assuming d and d2 are defined DataFrames
d = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
#Creates a DataFrame named 'd' with columns A and B
#AP24110010909

d2 = pd.DataFrame({'A': [7, 8, 9], 'B': [10, 11, 12]})
#Creates another DataFrame named 'd2' with columns A and B
#AP24110010909

# Replace append with concat
d = pd.concat([d, d2])
#Combines the DataFrames 'd' and 'd2' vertically using the concat() function
#AP24110010909

print(d)
#Displays the combined DataFrame on the screen
#AP24110010909
```

```python
#Deletion of rows:

import pandas as pd
#Imports the Pandas library and gives it the alias 'pd'
#AP24110010909

a_info = pd.DataFrame([[4, 5], [6, 7]], columns = ['x','y'])
#Creates a DataFrame named 'a_info' with columns x and y
#AP24110010909

b_info = pd.DataFrame([[8, 9], [10, 11]], columns = ['x','y'])
#Creates another DataFrame named 'b_info' with columns x and y
#AP24110010909

a_info = pd.concat([a_info,b_info])
#Combines the two DataFrames vertically using the concat() function
#AP24110010909

print(a_info)
#Displays the combined DataFrame on the screen
#AP24110010909

# Drop rows with label 0
a_info = a_info.drop(0)
#Deletes all rows with index label 0 from the DataFrame
#AP24110010909

print("Drop the row with label 0")
#Displays a message indicating that the row with label 0 is being deleted
#AP24110010909

print(a_info)
#Displays the DataFrame after deleting the rows with label 0
#AP24110010909
```

```python
#DataFrame.drop_duplicates()

import pandas as pd
#Imports the Pandas library and gives it the alias 'pd'
#AP24110010909

emp = {"Name": ["Siva", "Rama", "Krishna", "Siva"],
       "Age": [21, 32, 29, 21]}
#Creates a dictionary containing employee names and ages
#AP24110010909

info = pd.DataFrame(emp)
#Converts the dictionary into a Pandas DataFrame
#AP24110010909

print(info)
#Displays the original DataFrame on the screen
#AP24110010909

info = info.drop_duplicates()
#Removes duplicate rows from the DataFrame
#AP24110010909

print("After removing the duplicates the dataframe is :")
#Displays a message indicating that duplicates have been removed
#AP24110010909

print(info)
#Displays the DataFrame after removing duplicate rows
#AP24110010909
```

```python
#Null values identification in given dataset

import pandas as pd
#Imports the Pandas library and gives it the alias 'pd'
#AP24110010909

import numpy as np
#Imports the NumPy library and gives it the alias 'np'
#AP24110010909

info = pd.DataFrame({"Person":["Siva", "Pavan", "Gopal", "Raghu"],
                     "Age": [27., 29, np.nan, 32]})
#Creates a DataFrame with Person and Age columns
#np.nan represents a missing or null value
#AP24110010909

print(info)
#Displays the DataFrame on the screen
#AP24110010909

print(info.count())
#Counts and displays the number of non-null values in each column
#AP24110010909

print("null values")
#Displays a message indicating that null values are being identified
#AP24110010909

print(info.isnull())
#Checks each cell for null values
#True means the value is null and False means the value is not null
#AP24110010909

print(info.isnull().sum())
#Counts the total number of null values in each column
#AP24110010909
```

```python
# Sorting order

import pandas as pd
#Imports the Pandas library and gives it the alias 'pd'
#AP24110010909

import numpy as np
#Imports the NumPy library and gives it the alias 'np'
#AP24110010909

info = pd.DataFrame({"Person":["Sriu", "Ramu", "Mani", "Bhanu"],
                     "Age": [27., 29, np.nan, 32]})
#Creates a DataFrame with Person and Age columns
#np.nan represents a missing or null value
#AP24110010909

print(info)
#Displays the original DataFrame on the screen
#AP24110010909

print("Sorting Order")
#Displays a message indicating the sorting order
#AP24110010909

print(info.sort_index())
#Sorts the DataFrame according to the index in ascending order
#AP24110010909

print("Descending Order")
#Displays a message indicating descending order
#AP24110010909

print(info.sort_index(ascending=False))
#Sorts the DataFrame according to the index in descending order
#AP24110010909
```
